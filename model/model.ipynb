### Build a binary classification model for whatsapp messages, knowing if something is important or not. 

```python
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report
import joblib
```

```python
print('starting')
```

```python
with open('training_data/data.txt', 'r', encoding='utf-8') as file:
    messages = [line.strip() for line in file]

print(f"Loaded {len(messages)} messages")
```

```python
with open('training_data/labels.txt', 'r', encoding='utf-8') as file:
    labels = [int(line.strip()) for line in file]

print(f"Loaded {len(labels)} labels")
```

```python
model = Pipeline([
    ("tfidf", TfidfVectorizer(
        ngram_range=(1, 2),
        lowercase=True,
        stop_words="english"
    )),
    ("clf", LogisticRegression())
])
```

```python
model.fit(messages, labels)

# Save the trained model
joblib.dump(model, "message_model.joblib")

print("Model trained and saved.")
```

```python
with open('test_data/data.txt', 'r', encoding='utf-8') as file:
    test_messages = [line.strip() for line in file]

print(f"Loaded {len(test_messages)} test messages")
```

```python
with open('test_data/labels.txt', 'r', encoding='utf-8') as file:
    test_labels = [int(line.strip()) for line in file]

print(f"Loaded {len(test_labels)} test labels")
```

```python
# Load the model (optional, shows how to reload later)
loaded_model = joblib.load("message_model.joblib")

# Test the model
predictions = loaded_model.predict(test_messages)

# Evaluate performance
accuracy = accuracy_score(test_labels, predictions)
print("Accuracy:", accuracy)

print("\nClassification Report:")
print(classification_report(test_labels, predictions))
```
